```python
# Loss curves (ANN/LSTM): convergence & overfitting evidence
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, name in zip(axes, ['ann', 'lstm']):
    lc = pd.read_csv(art / f'{name}_loss_curve.csv')
    lc.plot(ax=ax); ax.set_title(f'{name.upper()} loss')
    ax.set_xlabel('epoch')
plt.tight_layout()
```
